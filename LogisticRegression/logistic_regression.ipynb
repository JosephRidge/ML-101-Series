# **Data Modeling**

```python
model = LogisticRegression(random_state= 2) # instance of the model
```

```python
model.fit(X, y) # process or training your model 
```

```python
model.predict(X[:2, :]) # running a prediction awaiting a class outcome
```

```python
model.predict_proba(X[:2, :]) # attain the probabilities (0-1)
```

```python
model.score(X, y)
```

```python
y_hat = model.predict(X)
print(len(y_hat)) # gets the size of the predictions
y_hat # gives us the entire otucomes i.e the predicted classes
```

probability:

```python
y_hat_probs = model.predict_proba(X) # attain hte probability values of each prediction
print(len(y_hat_probs)) 

#  formating the probabilities into 4 decimal places for easy readability
np.set_printoptions(precision = 4, suppress = True)
print(y_hat_probs)
```

```python
plt.figure(figsize=(8,5))

petal_len = X[:,2]

for i, class_name in enumerate(iris.target_names):
    sort_idx = np.argsort(petal_len)
    plt.plot(X.flatten()[sort_idx],  y_hat_probs[:,i][sort_idx], label=class_name)

plt.axhline(0.5, color ='gray', linestyle='--')
plt.xlabel('Petal Length')
plt.ylabel('Probability')
plt.legend()
plt.show()
```
